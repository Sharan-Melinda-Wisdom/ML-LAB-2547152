**Part B: Simple Linear Regression using Scikit-learn**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
df.columns = df.columns.str.strip()
print("Cleaned Columns in your Dataset:")
print(list(df.columns))
```

```python
target_columns=['your CIA % of last semester', 'Your maximum attendance % till last semester', 'your GPA of last semester']
for col in target_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df=df.dropna(subset=target_columns)

Y=df['your GPA of last semester']

print("\n"+"="*20+"EXPERIMENT 1: CIA % vs GPA"+"="*20)
X1=df[['your CIA % of last semester']] 
X1_train, X1_test, Y1_train, Y1_test = train_test_split(X1, Y, test_size=0.2, random_state=42)
model1=LinearRegression()
model1.fit(X1_train, Y1_train)
print(f"Slope (m): {model1.coef_[0]:.4f}")
print(f"Intercept (c): {model1.intercept_:.4f}")


print("\n" + "="*20 + " EXPERIMENT 2: Attendance % vs GPA " + "="*20)
X2=df[['Your maximum attendance % till last semester']]
X2_train, X2_test, Y2_train, Y2_test = train_test_split(X2, Y, test_size=0.2, random_state=42)
model2=LinearRegression()
model2.fit(X2_train, Y2_train)

print(f"Slope (m): {model2.coef_[0]:.4f}")
print(f"Intercept (c): {model2.intercept_:.4f}")
```

Experiment 1 (CIA to GPA): The slope is positive, indicating that GPA tends to increase as CIA percentage increases.

Experiment 2 (Attendance to GPA): The slope is positive, indicating that GPA tends to increase with attendance percentage.

**Part C: Manual Computation using Ordinary Least Squares (OLS)**

```python
import numpy as np

print("\n"+"="*20+"MANUAL OLS : CIA % vs GPA"+"="*20)

# Convert training data to arrays
x = X1_train['your CIA % of last semester']
y = Y1_train

# Mean values
x_mean = np.mean(x)
y_mean = np.mean(y)

# Slope calculation
m = np.sum((x-x_mean)*(y-y_mean)) / np.sum((x-x_mean)**2)

# Intercept calculation
b = y_mean - (m*x_mean)

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")

# Predictions
manual_pred1 = m * X1_test['your CIA % of last semester'] + b

print("\nPredicted GPA:")
print(manual_pred1)



print("\n"+"="*20+"MANUAL OLS : Attendance % vs GPA"+"="*20)

x = X2_train['Your maximum attendance % till last semester']
y = Y2_train

x_mean = np.mean(x)
y_mean = np.mean(y)

m = np.sum((x-x_mean)*(y-y_mean)) / np.sum((x-x_mean)**2)

b = y_mean - (m*x_mean)

print(f"Slope (m): {m:.4f}")
print(f"Intercept (b): {b:.4f}")

# Predictions
manual_pred2 = m * X2_test['Your maximum attendance % till last semester'] + b

print("\nPredicted GPA:")
print(manual_pred2)
```

**Comparison Task:**

1. Predictions from Scikit learn

The Scikit-learn Linear Regression model generated GPA prediction values using the slope and intercept obtained after training on the dataset.

2. Predictions from manually computed

The manual OLS method generated prediction values using the regression equation: Y=mX+b

3. Difference between both outputs

The predictions from Scikit-learn and manually computed OLS are identical. Both methods produced the same slope and intercept values because the same training dataset was used in both approaches

**Parameter Saving Task**

```python
import pickle

print("\n"+"="*20+" SAVING PARAMETERS "+"="*20)

# Store model parameters
parameters = {
    'slope': model1.coef_[0],
    'intercept': model1.intercept_
}

# Save into Pickle file
with open('linear_regression_weights.pkl','wb') as file:
    pickle.dump(parameters,file)

print("Parameters saved successfully!")
```

```python
print("\n"+"="*20+" LOADING PARAMETERS "+"="*20)

with open('linear_regression_weights.pkl','rb') as file:
    loaded_parameters=pickle.load(file)

print("Loaded Parameters:")
print(loaded_parameters)
```

```python
print("\n"+"="*20+" PREDICTION USING LOADED PARAMETERS "+"="*20)

slope=loaded_parameters['slope']
intercept=loaded_parameters['intercept']

new_CIA=85

predicted_GPA=(slope*new_CIA)+intercept

print("CIA Percentage:",new_CIA)
print("Predicted GPA:",predicted_GPA)
```

The model parameters are successfully stored using a Pickle file. The saved parameters are then loaded back & used to predict GPA values without retraining the model. This reduces computation time and allows reuse of the trained model.
